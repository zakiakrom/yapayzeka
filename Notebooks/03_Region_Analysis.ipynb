```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Load data bersih
df = pd.read_csv(r'D:\finish\New folder\Kuliah\Semester 6\Yapay Zeka\yapayzeka\Data\Processed\ecommerce_clean.csv')

# 2. Grouping: Total Sales per Region dan Category
region_sales = df.groupby(['region', 'category'])['sales'].sum().reset_index()

# 3. Visualisasi dengan Bar Chart
plt.figure(figsize=(12, 6))
sns.barplot(data=region_sales, x='region', y='sales', hue='category')
plt.title("Distribution of Sales by Region and Category")
plt.xlabel("Region")
plt.ylabel("Total Sales ($)")
plt.legend(title="Product Category", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
